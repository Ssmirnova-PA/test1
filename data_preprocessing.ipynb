## Предобработка данных для кейса "Аналитика маркетплейса: поиск причин падения выручки, юнит-экономика и A/B-тест"

Этот блок содержит загрузку, проверку и очистку исходных данных: таблиц пользователей, событий, заказов и маркетинговых затрат. Очищенные данные используются в основном проекте, где выполняется анализ метрик, юнит-экономики и пользовательской воронки

```python
import pandas as pd
```

```python
orders_raw = pd.read_csv('https://..../datasets/pa_marketplace_orders.csv',parse_dates=['order_date', 'order_week', 'order_month'])
campaign_costs = pd.read_csv('https://..../datasets/pa_marketplace_campaign_costs.csv', parse_dates=['spend_month'])
users = pd.read_csv('https://code.s3.yandex.net/datasets/pa_marketplace_users.csv',parse_dates=['registration_date','cohort_week','cohort_month'])
events_raw = pd.read_csv('https://code.s3.yandex.net/datasets/pa_marketplace_events.csv',parse_dates=['event_date','event_week','event_month'])
```

```python
# Создаем словарь датафреймов с вашими названиями
tables = {
    'orders_raw': orders_raw,
    'campaign_costs': campaign_costs,
    'users': users,
    'events_raw': events_raw
}

# Создаем функцию для вывода краткой информации
def dataframe_summary(name, df):
    summary = pd.DataFrame({
        'column': df.columns,
        'dtype': df.dtypes.values,
        'non_null': df.count().values,
        'nulls': df.isnull().sum().values,
        'nulls_%': (df.isnull().sum() / len(df) * 100).round(2).values,
        'unique': df.nunique().values
    })
    return summary

# Выводим информацию по датафреймам
for name, df in tables.items():
    print(f'Датафрейм: {name} (строк - {len(df)}, колонок - {len(df.columns)})')
    display(dataframe_summary(name, df))
    display(df.head())
    print()
```

Пропуски: в orders_raw, campaign_costs и users пропусков нет. В events_raw пропущено 400 119 значений product_name (50,91%). Пропуски product_name в данных обусловлены структурой данных, так как в event_type содержатся разные события, в том числе не связаные с покупками

Названия столбцов: соответствуют стилю snake_case

Типы данных: в целом определены корректно: идентификаторы и количественные показатели представлены числовыми типами, даты — datetime64, категориальные признаки — строковыми типами.

Предварительная оценка: критических проблем с пропусками и типами данных не обнаружено.

## Дубликаты

```python
#Проверяю явные дубликаты 
dup_orders = orders_raw.duplicated().sum()
dup_campaign_costs = campaign_costs.duplicated().sum()
dup_users = users.duplicated().sum()
dup_events = events_raw.duplicated().sum()

print(f"Количество дубликатов по всему df orders : {dup_orders}")
print(f"Количество дубликатов по всему df campaign_costs : {dup_campaign_costs}")
print(f"Количество дубликатов по всему df users : {dup_users}")
print(f"Количество дубликатов по всему df events  : {dup_events}")
```

Явные дубликаты отсутствуют 

```python
# Проверка неявных дубликатов

# --- Users ---
duplicate_cols_users = ['user_id', 'registration_date', 'region', 'acq_channel']
duplicates_count = users.duplicated(subset=duplicate_cols_users).sum()
print(f"Users | Найдено неявных дубликатов по полям {duplicate_cols_users}: {duplicates_count}")

# --- Events ---
duplicate_cols_events = ['user_id', 'event_date', 'event_type', 'device']
duplicates_count = events_raw.duplicated(subset=duplicate_cols_events).sum()
print(f"Events | Найдено неявных дубликатов по полям {duplicate_cols_events}: {duplicates_count}")

# --- Orders ---
duplicate_cols_orders = ['user_id', 'order_date', 'product_name', 'quantity']
duplicates_count = orders_raw.duplicated(subset=duplicate_cols_orders).sum()
print(f"Orders | Найдено неявных дубликатов по полям {duplicate_cols_orders}: {duplicates_count}")

# --- Campaign_costs ---
duplicate_cols_campaigns = ['acq_channel', 'spend_month']
duplicates_count = campaign_costs.duplicated(subset=duplicate_cols_campaigns).sum()
print(f"Campaign_costs | Найдено неявных дубликатов по полям {duplicate_cols_campaigns}: {duplicates_count}")
```

```python
# Чекаю примеры неявных дубликатов

#  Events
duplicate_cols_events = ['user_id', 'event_date', 'event_type', 'device']
events_duplicates = events_raw[events_raw.duplicated(subset=duplicate_cols_events, keep=False)]
events_duplicates_sorted = events_duplicates.sort_values(by=duplicate_cols_events)
print(f"Events | Примеры неявных дубликатов ({len(events_duplicates)} строк):")
display(events_duplicates_sorted.head(10))

print()

#  Orders
duplicate_cols_orders = ['user_id', 'order_date', 'product_name', 'quantity']
orders_duplicates = orders_raw[orders_raw.duplicated(subset=duplicate_cols_orders, keep=False)]
orders_duplicates_sorted = orders_duplicates.sort_values(by=duplicate_cols_orders)
print(f"Orders | Примеры неявных дубликатов ({len(orders_duplicates)} строк):")
display(orders_duplicates_sorted.head(10))
```

Events - Удаляем. Одно и то же событие (например, checkout_complete) не может срабатывать множество раз в один день на одном устройстве у одного пользователя.

Orders —  Удаляем. Один и тот же товар, в одном количестве, по одной цене, оформленный одним пользователем в одну и ту же секунду — это двойная запись на стороне бэкенда, а не два реальных заказа. Удаление не искажает выручку.

```python
#Дропаю дубликаы и перезаписываю очищенные данные в новый дф
#events
events_before = len(events_raw)
duplicate_cols_events = ['user_id', 'event_date', 'event_type', 'device', 'os']
events = events_raw.drop_duplicates(subset=duplicate_cols_events, keep='first')
events_removed = events_before - len(events)
print(f"Events | Было: {events_before}, удалено: {events_removed} ({events_removed / events_before * 100:.2f}%), осталось: {len(events)}")

# orders
orders_before = len(orders_raw)
duplicate_cols_orders = ['user_id', 'order_date', 'product_name', 'quantity', 'unit_price']
orders = orders_raw.drop_duplicates(subset=duplicate_cols_orders, keep='first')
orders_removed = orders_before - len(orders)
print(f"Orders | Было: {orders_before}, удалено: {orders_removed} ({orders_removed / orders_before * 100:.2f}%), осталось: {len(orders)}")
```
