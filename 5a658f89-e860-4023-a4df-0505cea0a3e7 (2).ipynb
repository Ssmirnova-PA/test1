Events — Удаляем. Одно и то же событие (например, checkout_complete) не может срабатывать множество раз в один день на одном устройстве у одного пользователя — это баг трекинга (повторные клики по кнопке из-за лага или повторное срабатывание кода аналитики при обновлении страницы + NaN в product_name ).

Orders —  Удаляем. Один и тот же товар, в одном количестве, по одной цене, оформленный одним пользователем в одну и ту же секунду — это двойная запись на стороне бэкенда, а не два реальных заказа. Удаление не искажает выручку.

````python
#Дропаем дубликаы
#events
events_before = len(events)
duplicate_cols_events = ['user_id', 'event_date', 'event_type', 'device', 'os']
events = events.drop_duplicates(subset=duplicate_cols_events, keep='first')
events_removed = events_before - len(events)
print(f"Events | Было: {events_before}, удалено: {events_removed} ({events_removed / events_before * 100:.2f}%), осталось: {len(events)}")

# orders
orders_before = len(orders)
duplicate_cols_orders = ['user_id', 'order_date', 'product_name', 'quantity', 'unit_price']
orders = orders.drop_duplicates(subset=duplicate_cols_orders, keep='first')
orders_removed = orders_before - len(orders)
print(f"Orders | Было: {orders_before}, удалено: {orders_removed} ({orders_removed / orders_before * 100:.2f}%), осталось: {len(orders)}")
````

````python
#Расчет ретенш, сначала считаем дату первого заказа 
orders['first_order_month'] = orders.groupby('user_id')['order_month'].transform('min')

orders['n_period'] = (
    orders['order_month'].dt.to_period('M').astype('int64')
    - orders['first_order_month'].dt.to_period('M').astype('int64')
)


data_pivot = pd.pivot_table(orders,
                            values='user_id',
                            index='first_order_month',
                            columns='n_period',
                            aggfunc=pd.Series.nunique).fillna(0).astype(int)



# Каждое значение строки датафрейма делим на значение в периоде 0 с помощью функции div()
data_pivot = data_pivot.div(data_pivot[0], axis=0)

# Заменяем нули на пустые значения NaN. 
data_pivot[data_pivot == 0] = float('nan')

cmap = mcolors.LinearSegmentedColormap.from_list('green_red', ['#FF3333', '#FFFF66', '#4C9900'], N=100) # красивые цвета heatmap

# Для корректного отображения индекса приведём его к типу ('%Y-%m-%d')
data_pivot.index = pd.to_datetime(data_pivot.index).strftime('%Y-%m-%d')

plt.subplots(figsize=(20,12))
sns.heatmap(data_pivot,
           cmap=cmap,
           annot=True,
           fmt='.0%',
           linewidth=0.5,
           annot_kws={"size": 18})

plt.title('N-Period Retention по когортам в %')
plt.xlabel('Периоды')
plt.ylabel('Когорты')
plt.yticks(rotation=0)
plt.tick_params(labelsize=12) 
plt.show()
````

````python
# Считая САС берем именно платящих пользователей, Считаем выручку маркетплейса (5% комиссии от итоговой суммы заказа total_price)
orders['revenue'] = orders['total_price'] * 0.05

#  Подтягиваем к заказам канал привлечения (acq_channel) из таблицы Users
df_merged = orders.merge(
    users[['user_id', 'acq_channel']], 
    on='user_id', 
    how='left'
)

#  Рассчитываем номер периода N_month (0, 1, 2...) простым вычитанием месяцев
df_merged['N_month'] = df_merged['order_month'].dt.month - df_merged['first_order_month'].dt.month

#  Агрегируем выручку до уровня пользователя и его месяца жизни
df_LTV = df_merged.groupby(
    ['user_id', 'acq_channel', 'first_order_month', 'N_month']
)['revenue'].sum().reset_index()

#  Сортируем по N_month для корректной кумулятивной суммы
df_LTV = df_LTV.sort_values(by=['user_id', 'acq_channel', 'N_month'])

#  Считаем кумулятивный LTV для каждого пользователя при помощи cumsum()
df_LTV['LTV'] = df_LTV.groupby(['user_id'])['revenue'].cumsum()

# Проверяем результат
print(df_LTV.head())
````

<div class="alert alert-success">
<b>Комментарий ревьюера v1:</b>

Расчет кумулятивного LTV через `cumsum()` выполнен абсолютно корректно. Логика группировки по месяцу жизни когорты правильная.
</div>

````python
# 1. Считаем количество уникальных платящих клиентов для каждой пары (acq_channel, first_order_month)
# Мы берём их прямо из твоего df_LTV, так как там лежат только реальные покупатели
client_counts = df_LTV.groupby(['acq_channel', 'first_order_month'])['user_id'].nunique().reset_index()
client_counts.columns = ['acq_channel', 'first_order_month', 'unique_customers']

# 2. Объединяем маркетинговые бюджеты с количеством клиентов
# Мержим по каналу и месяцу (spend_month в бюджетах равен first_order_month у клиентов)
cac_data = campaign_costs.merge(
    client_counts, 
    left_on=['spend_month', 'acq_channel'], 
    right_on=['first_order_month', 'acq_channel'],
    how='left'
)

# Заполняем пропуски в количестве клиентов нулями (на случай, если где-то не было продаж)
cac_data['unique_customers'] = cac_data['unique_customers'].fillna(0)

# 3. Рассчитываем CAC (бюджет / количество платящих клиентов)
cac_data['CAC'] = cac_data['budget'] / cac_data['unique_customers']

# Если клиентов было 0, заменяем получившуюся бесконечность (inf) на 0
cac_data['CAC'] = cac_data['CAC'].replace([np.inf, -np.inf], 0)

# 4. Присоединяем значение CAC по месяцу и каналу к твоему исходному df_LTV
# Результат сохраняем в новую таблицу df_LTV_CAC, как ты и просила
df_LTV_CAC = df_LTV.merge(
    cac_data[['first_order_month', 'acq_channel', 'CAC']], 
    on=['first_order_month', 'acq_channel'], 
    how='left'
)

display(df_LTV_CAC.head(5))
````

````python
# Задаем порядок шагов в воронке
funnel_steps = ['product_view', 'add_to_cart', 'checkout_start', 'checkout_complete']

# Фильтруем датасет оставляя только нужные шаги воронки
funnel_df_steps = events[events['event_type'].isin(funnel_steps)]

# Считаем число уникальных пользователей на каждом шаге
funnel_counts = funnel_df_steps.groupby('event_type')['user_id'].nunique().reindex(funnel_steps)

# Добавляем реальные покупки из Orders как финальный шаг
orders_users = orders['user_id'].nunique()
funnel_steps_full = funnel_steps + ['order_placed']
funnel_counts_full = list(funnel_counts.values) + [orders_users]

# Собираем таблицу воронки
funnel_df = pd.DataFrame({
    'step': funnel_steps_full,
    'users': funnel_counts_full
})

# Добавляем конверсии
funnel_df['conversion_from_first_%'] = (funnel_df['users'] / funnel_df['users'].iloc[0] * 100).round(1)
funnel_df['conversion_from_previous_%'] = (funnel_df['users'] / funnel_df['users'].shift(1) * 100).round(1)

display(funnel_df)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# График 1 — все шаги включая product_view (100%)
steps_first = funnel_df['step']
values_first = funnel_df['conversion_from_first_%']

# График 2 — без первого шага (у него нет предыдущего)
steps_prev = funnel_df['step'][1:]
values_prev = funnel_df['conversion_from_previous_%'][1:]

# ГРАФИК 1: Общая воронка 
ax1.bar(steps_first, values_first, color='gray')
ax1.set_title('Общая воронка\n(доля от шага product_view)', fontsize=12)
ax1.set_ylabel('Конверсия (%)')
ax1.set_ylim(0, max(values_first) + 10)
ax1.tick_params(axis='x', rotation=45)
for i, val in enumerate(values_first):
    ax1.text(i, val + 1, f'{val}%', ha='center', fontweight='bold')

# ГРАФИК 2: Step-by-step воронка 
ax2.bar(steps_prev, values_prev, color='orange')
ax2.set_title('Step-by-step воронка', fontsize=12)
ax2.set_ylabel('Конверсия (%)')
ax2.set_ylim(0, max(values_prev) + 10)
ax2.tick_params(axis='x', rotation=45)
for i, val in enumerate(values_prev):
    ax2.text(i, val + 1, f'{val}%', ha='center', fontweight='bold')

plt.tight_layout()
plt.show()
````

Рассмотрим воронки в развере устройств

````python
# Задаем порядок шагов в воронке
funnel_steps = ['product_view', 'add_to_cart', 'checkout_start', 'checkout_complete']
devices = events['device'].unique()

for device in devices:
    # Фильтруем events по устройству
    events_device = events[events['device'] == device]
    
    # Фильтруем оставляя только нужные шаги воронки
    funnel_df_steps = events_device[events_device['event_type'].isin(funnel_steps)]
    
    # Считаем число уникальных пользователей на каждом шаге
    funnel_counts = funnel_df_steps.groupby('event_type')['user_id'].nunique().reindex(funnel_steps)
    
    # Добавляем реальные покупки из Orders как финальный шаг
    # Берём только пользователей которые заходили с этого устройства
    device_users = events_device['user_id'].unique()
    orders_users = orders[orders['user_id'].isin(device_users)]['user_id'].nunique()
    
    funnel_steps_full = funnel_steps + ['order_placed']
    funnel_counts_full = list(funnel_counts.values) + [orders_users]
    
    # Собираем таблицу воронки
    funnel_df = pd.DataFrame({
        'step': funnel_steps_full,
        'users': funnel_counts_full
    })
    
    # Добавляем конверсии
    funnel_df['conversion_from_first_%'] = (funnel_df['users'] / funnel_df['users'].iloc[0] * 100).round(1)
    funnel_df['conversion_from_previous_%'] = (funnel_df['users'] / funnel_df['users'].shift(1) * 100).round(1)
    
    print(f'\n=== Устройство: {device} ===')
    display(funnel_df)
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))
    fig.suptitle(f'Воронка для устройства: {device}', fontsize=14, fontweight='bold')
    
    # График 1 — все шаги включая product_view (100%)
    steps_first = funnel_df['step']
    values_first = funnel_df['conversion_from_first_%']
    
    # График 2 — без первого шага (у него нет предыдущего)
    steps_prev = funnel_df['step'][1:]
    values_prev = funnel_df['conversion_from_previous_%'][1:]
    
    # ГРАФИК 1: Общая воронка
    ax1.bar(steps_first, values_first, color='gray')
    ax1.set_title('Общая воронка\n(доля от шага product_view)', fontsize=12)
    ax1.set_ylabel('Конверсия (%)')
    ax1.set_ylim(0, max(values_first) + 10)
    ax1.tick_params(axis='x', rotation=45)
    for i, val in enumerate(values_first):
        ax1.text(i, val + 1, f'{val}%', ha='center', fontweight='bold')
    
    # ГРАФИК 2: Step-by-step воронка
    ax2.bar(steps_prev, values_prev, color='orange')
    ax2.set_title('Step-by-step воронка', fontsize=12)
    ax2.set_ylabel('Конверсия (%)')
    ax2.set_ylim(0, max(values_prev) + 10)
    ax2.tick_params(axis='x', rotation=45)
    for i, val in enumerate(values_prev):
        ax2.text(i, val + 1, f'{val}%', ha='center', fontweight='bold')
    
    plt.tight_layout()
    plt.show()
````

Все три устройства показывают идентичную картину — проблема не в устройстве, а в самом шаге оплаты. Последний переход checkout_complete → order_placed проседает на всех платформах: mobile 54.1%, desktop 50.9%, tablet 52.5%. Это системная проблема продукта, а не UX конкретного устройства.

Если это подтверждается, то можно сделать упор на определенный канал для повышения выручки и наращивания базы платежеспособных пользователей 

````python
# Создаем таблицу с базовой статистикой пользователей по сегментам
buyer_segment_info = users.groupby('buyer_segment').agg(
    {'user_id': 'count'}
).reset_index().rename(columns={
    'buyer_segment': 'Сегмент покупателя', 
    'user_id': 'Кол-во пользователей'
})

# Добавляем долю пользователей в процентах
total_users = buyer_segment_info['Кол-во пользователей'].sum()
buyer_segment_info['Доля, %'] = (buyer_segment_info['Кол-во пользователей'] / total_users * 100).round(2)

#  Готовим данные для конверсии: находим уникальных покупателей по сегментам
# Берем всех, кто есть в orders (уникальные user_id)
paid_users_per_segment = orders.merge(users[['user_id', 'buyer_segment']], on='user_id')['user_id'].unique()

# Считаем количество платящих пользователей в каждом сегменте
paid_stats = users.copy()
paid_stats['is_paid'] = paid_stats['user_id'].isin(paid_users_per_segment).astype(int)
cr_data = paid_stats.groupby('buyer_segment')['is_paid'].agg(['sum', 'count']).reset_index()

#  Считаем LTV (средняя выручка на ВСЕХ пользователей в сегменте, даже если 0)
revenue_per_user = orders.groupby('user_id')['revenue'].sum().reset_index()
ltv_data = users[['user_id', 'buyer_segment']].merge(revenue_per_user, on='user_id', how='left')
ltv_data['revenue'] = ltv_data['revenue'].fillna(0)
ltv_stats = ltv_data.groupby('buyer_segment')['revenue'].mean().reset_index()

#  Объединяем всё в одну таблицу
final_df = buyer_segment_info.copy()

# Добавляем CR: переименуем столбец в cr_data прямо перед мерджем, чтобы не плодить дубли
final_df = final_df.merge(cr_data.rename(columns={'buyer_segment': 'Сегмент покупателя'}), on='Сегмент покупателя')
final_df['Конверсия, %'] = (final_df['sum'] / final_df['count'] * 100).round(2)

# Добавляем LTV: также переименуем перед мерджем
final_df = final_df.merge(ltv_stats.rename(columns={'buyer_segment': 'Сегмент покупателя'}), on='Сегмент покупателя')

# Переименовываем выручку в LTV (округляем только этот столбец, а не всю таблицу)
final_df = final_df.rename(columns={'revenue': 'Средний LTV (руб.)'})
final_df['Средний LTV (руб.)'] = final_df['Средний LTV (руб.)'].round(2)

#  Оставляем только нужные столбцы
final_table = final_df[['Сегмент покупателя', 'Кол-во пользователей', 'Доля, %', 'Конверсия, %', 'Средний LTV (руб.)']]

display(final_table)
````

````python
# RFM анализ

#  Рассчитываем Recency с учетом времени последнего заказа ( чтобы вошли заказы за все время 31 декабря)
analysis_date = pd.to_datetime('2024-12-31 23:59:59')
orders['order_recency'] = analysis_date - orders['order_date']

#  Группируем данные (теперь в order_recency корректные значения)
ppro_rfm = orders.groupby('user_id').agg(
    # Берем min(), так как чем меньше дней, тем свежее покупка
    recency = ('order_recency', lambda x: x.min().days),
    frequency = ('order_id', 'count'),
    monetary_value = ('revenue', 'sum')
).reset_index()

print(ppro_rfm.head(5))

fig, axes = plt.subplots(3, 1, figsize=(15, 15))


# Давность
axes[0].hist(ppro_rfm['recency'], bins=75,color='gray')
axes[0].set_title('Recency (дни с последней покупки)')
axes[0].set_xlabel('Дни')
axes[0].set_ylabel('Количество пользователей')

# Частота
axes[1].hist(ppro_rfm['frequency'], bins=75,color='gray')
axes[1].set_title('Frequency (число заказов)')
axes[1].set_xlabel('Количество заказов')
axes[1].set_ylabel('Количество пользователей')

# Сумма всех заказов
axes[2].hist(ppro_rfm['monetary_value'], bins=75,color='gray')
axes[2].set_title('Monetary (стоимость заказов)')
axes[2].set_xlabel('Стоимость заказов')
axes[2].set_ylabel('Количество пользователей')
````

<div class="alert alert-success">
<b>Комментарий ревьюера v1:</b>

Реализация RFM-анализа выполнена на высоком уровне. Использование `pd.qcut` и `pd.cut` для сегментации — правильный подход.
</div>

````python
# Определяем группу пользователя по количеству дней с момента последнего заказа: 
ppro_rfm['r'] = pd.cut(ppro_rfm['recency'], bins=3, labels=[3,2,1])

# Определяем группу пользователя по количеству заказов: 
ppro_rfm['f'] = pd.cut(ppro_rfm['frequency'], bins=[0, 1, 2, float('inf')], labels=[1, 2, 3], include_lowest=True)

# Определяем группу пользователя по сумме стоимости заказов: 
ppro_rfm['m'] = pd.qcut(ppro_rfm['monetary_value'], q=3, labels=[1, 2, 3])
# Находим групповой RFM-индекс:
ppro_rfm[['r','f','m']] = ppro_rfm[['r','f','m']].astype('str')
ppro_rfm['rfm_group'] = ppro_rfm['r'] + ppro_rfm['f'] + ppro_rfm['m']

# Выводим первые строки таблицы:
print(ppro_rfm.head(3))
````

````python
# 1. Группируем и считаем показатели
rfm_group = ppro_rfm.groupby('rfm_group').agg({'user_id':'nunique'}).reset_index()
rfm_group['user_share'] = (rfm_group['user_id'] / rfm_group['user_id'].sum() * 100).round(1)

# 2. Создаем стилизованную таблицу
styled_rfm = rfm_group.sort_values(by='user_id', ascending=False).style.background_gradient(
    cmap='Blues',           
    subset=['user_id', 'user_share']  
)

#Выводим результат
styled_rfm
````

## Категории товаров

````python
# 1. Группируем заказы по категориям и считаем метрики
category_analysis = orders.groupby('category_name').agg({
    'order_id': 'nunique',       # Частота (количество уникальных заказов с этой категорией)
    'quantity': 'sum',           # Объем (сколько всего единиц продано)
    'total_price': ['mean', 'sum'] # Средний чек и общая выручка
}).reset_index()

# 2. Упрощаем структуру колонок после .agg()
category_analysis.columns = [
    'Категория', 
    'Кол-во заказов', 
    'Кол-во товаров', 
    'Средний чек (руб.)', 
    'Выручка (руб.)'
]

# 3. Округляем денежные показатели
category_analysis['Средний чек (руб.)'] = category_analysis['Средний чек (руб.)'].round(2)
category_analysis['Выручка (руб.)'] = category_analysis['Выручка (руб.)'].round(2)

# 4. Сортируем по выручке, чтобы сразу видеть топ-категории
category_analysis = category_analysis.sort_values(by='Выручка (руб.)', ascending=False)

# Вывод таблицы
display(category_analysis)
````

# Дизайн эксперимента:

Бонус на первую покупку для пользователей TikTok

## 1. Контекст и цель эксперимента

- **Бизнес-контекст:** Команда продукта запускает эксперимент, направленный на повышение конверсии среди пользователей, привлечённых через канал TikTok. Для стимулирования первой покупки пользователям будет предоставлен бонус на первую покупку.  
- **Цель:** Проверить, влияет ли предоставление бонуса на первую покупку на конверсию и выручку среди пользователей TikTok.  
- **Гипотеза:** Предоставление бонуса на первую покупку новым пользователям, привлеченным через TikTok, приведет к увеличению конверсии в первую покупку

---

## 2. Формат эксперимента

- **Тип эксперимента:** Сплит-роллинг (rolling split)  
- **Канал:** Только пользователи, пришедшие с TikTok  
- **Группы:**  
  - **Контрольная (A):** пользователи TikTok без бонуса  
  - **Тестовая (B):** пользователи TikTok с бонусом на первую покупку  
- **Распределение:** Равномерное распределение новых пользователей TikTok между группами по дате регистрации  
- **Период проведения:** 3 недели (фиксировано)

---

## 3. Ключевые метрики

### Основная метрика

- **Конверсия в первую покупку** — доля пользователей, зарегистрировавшихся с TikTok и совершивших первую покупку  
  - *Цель:* увеличить конверсию с помощью бонуса  
  - *Критерий успеха:* статистически значимый рост конверсии в тестовой группе

### Вспомогательные метрики

- **ARPU** (средняя выручка на пользователя TikTok)  
- **ARPPU** (средняя выручка на покупателя TikTok)  
- **AOV** (средний чек TikTok)

### Барьерная метрика

- **Общая выручка от пользователей TikTok**

---

## 4. Размеры групп и запуск

- **Целевая аудитория:** только пользователи TikTok  
- **Размеры групп:** определяются по фактическому трафику TikTok за период эксперимента  
- **Комментарий:** запуск без предварительного расчёта мощности и MDE из-за срочности задачи

## Основная метрика — конверсия в первую покупку


<div style="background-color: #d4edda; color: #155724; padding: 15px; border-radius: 4px; border: 1px solid #c3e6cb;">

<b>Комментарий ревьюера v2:</b>

💡 Не забывай, пожалуйста, формулировать нулевую и альтернативную гипотезы до проведения теста.



````python
# Получаем уникальные заказы из orders_AB
# Сортируем и берем первую покупку для каждого юзера
first_orders = (orders_AB.sort_values('order_date')
                .drop_duplicates(subset=['user_id'], keep='first'))

#  Соединяем  с первой покупкой
conversion_data = users_with_groups.merge(
    first_orders[['user_id', 'order_id', 'order_date', 'total_price']], 
    on='user_id', 
    how='left'
)

# Создаем флаг покупки (1 - купил, 0 - нет)
conversion_data['is_purchased'] = conversion_data['order_id'].notna().astype(int)


cr_by_group = conversion_data.groupby('group')['is_purchased'].mean() * 100


control_cr = cr_by_group['control']
test_cr = cr_by_group['treatment']

main_metric_result = pd.DataFrame({
    'Метрика': ['CR в первую покупку (%)'],
    'Control': [control_cr.round(2)],
    'Test': [test_cr.round(2)],
    'Абс. изменение (п.п.)': [(test_cr - control_cr).round(2)],
    'Отн. изменение (%)': [((test_cr - control_cr) / control_cr * 100).round(2)]
})

display(main_metric_result)
````

Конверсия в первую покупку в тестовой группе составила 7.51% против 1.06% в контроле — разница 6.44 п.п. в пользу теста, относительный рост 605% что вызвает сомнения в достоверности данных 

````python
# 1. Расчет Z-теста
control_data = conversion_data[conversion_data['group'] == 'control']['is_purchased']
treatment_data = conversion_data[conversion_data['group'] == 'treatment']['is_purchased']

n_control = len(control_data)
n_treatment = len(treatment_data)
conv_control = control_data.mean() * 100
conv_treatment = treatment_data.mean() * 100

# Количество успехов (покупок)
successes = [control_data.sum(), treatment_data.sum()]
# Общее количество наблюдений
nobs = [n_control, n_treatment]

z_stat, p_value = proportions_ztest(successes, nobs=nobs, alternative='two-sided')

print(f"Конверсия Control: {conv_control:.2f}%")
print(f"Конверсия Treatment: {conv_treatment:.2f}%")
print(f"Z-статистика: {z_stat:.4f}")
print(f"P-value: {p_value:.4f}")
print(f"Результат значимости: {'Статистически значимо' if p_value < 0.05 else 'Статистически НЕ значимо'} (α=0.05)")
````

Разница в конверсии статистически значима: z-статистика = −8.71, p-value < 0.001. Это означает, что вероятность получить такой результат случайно, при условии что реального эффекта нет, практически нулевая. Бонус значимо повышает конверсию в первую покупку.

Однако формальная значимость не отменяет ограничений, описанных ранее: малый абсолютный размер выборки (16 покупок в контроле) и возможное влияние сезонности. Статистический тест отвечает на вопрос «есть ли разница», но не гарантирует, что эффект воспроизведётся в других условиях.

````python
# Все участники теста 
all_users_registry = users_with_groups[['user_id', 'group']].drop_duplicates()

# Первые покупки
first_purchases = (orders_AB[orders_AB['user_id'].isin(users_with_groups['user_id'])]
                   .sort_values('order_date')
                   .drop_duplicates(subset=['user_id'], keep='first')
                   [['user_id', 'order_date']])

# Присоединяем покупки ко ВСЕМ пользователям
base_data = all_users_registry.merge(first_purchases, on='user_id', how='left')
base_data['is_purchased'] = base_data['order_date'].notna().astype(int)

# Идём по датам ПОКУПОК (а не регистраций)
purchase_dates = sorted(first_purchases['order_date'].dt.date.unique())

pval_records = []

for current_date in purchase_dates:
    
    # Знаменатель ВСЕГДА полный
    # Числитель: те, чья дата покупки <= current_date
    base_data['purchased_by_date'] = (
        base_data['order_date'].notna() & 
        (base_data['order_date'].dt.date <= current_date)
    ).astype(int)
    
    control = base_data[base_data['group'] == 'control']['purchased_by_date']
    treatment = base_data[base_data['group'] == 'treatment']['purchased_by_date']
    
    # Проверка, что есть покупки в обеих группах
    if control.sum() == 0 and treatment.sum() == 0:
        continue
    
    successes = [control.sum(), treatment.sum()]
    nobs = [len(control), len(treatment)]
    
    _, p_val = proportions_ztest(successes, nobs=nobs, alternative='two-sided')
    
    pval_records.append({
        'date': current_date,
        'p_value': p_val,
        'control_cr': control.mean() * 100,
        'treatment_cr': treatment.mean() * 100
    })

pvalue_df = pd.DataFrame(pval_records)

# График
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8))

ax1.plot(pvalue_df['date'], pvalue_df['p_value'], 
         color='steelblue', linewidth=2, marker='o')
ax1.axhline(y=0.05, linestyle='--', color='red', label='α=0.05')
ax1.set_title('Стабильность p-value по дням')
ax1.set_ylabel('p-value')
ax1.grid(True, alpha=0.3)
ax1.legend()

ax2.plot(pvalue_df['date'], pvalue_df['control_cr'], 
         label='Control', color='blue')
ax2.plot(pvalue_df['date'], pvalue_df['treatment_cr'], 
         label='Treatment', color='orange')
ax2.set_title('Накопленная конверсия')
ax2.set_ylabel('Конверсия, %')
ax2.grid(True, alpha=0.3)
ax2.legend()

plt.tight_layout()
plt.show()
````

P-value пересёк порог α=0.05 примерно на 7-й день эксперимента (около 7 января) и с тех пор устойчиво держится вблизи нуля — это говорит о том, что различие между группами статистически значимо и стабильно. Накопленная конверсия Treatment к концу периода достигла ~7.4% против ~1.1% в Control — разрыв более чем в 6 раз. При этом конверсия Treatment растёт равномерно на протяжении всего периода без резких скачков, что исключает эффект новизны как объяснение результата. Однако стоит отметить что данные начинаются с 1 января, и в первые 3–4 дня p-value был нестабилен и колебался от 0.1 до 0.3 — веротяно,что покупок ещё почти не было. Пользователи только зарегистрировались, но не успели совершить первую покупку — числитель был близок к нулю в обеих группах, и любая единичная покупка сильно меняла соотношение

Теперь нужно понять, не съедает ли бонус всю прибыль. Высокая конверсия может не окупаться.

💡Для анализа конверсии помимо Z-test можно также использовать альтернативные методы:
    
- t-test с линеаризацией: Можно преобразовать бинарные данные о покупках (0/1 для каждого пользователя) и применить обычный t-test к этим значениям. Поскольку при больших выборках результат будет аналогичен Z-test, это валидная альтернатива.
    
- Бутстрап: Создать множество ресемплированных выборок из данных и сравнить распределения конверсии между группами. Этот метод не требует предположений о распределении и особенно полезен при небольших выборках или когда мы хотим получить доверительные интервалы для разности конверсий.
    
```python
from scipy import stats
import numpy as np

def bootstrap_mean_diff(group1, group2, n_bootstrap=10000):
    differences = []
    for _ in range(n_bootstrap):
        sample1 = np.random.choice(group1, size=len(group1), replace=True)
        sample2 = np.random.choice(group2, size=len(group2), replace=True)
        differences.append(sample2.mean() - sample1.mean())
    
    ci_lower = np.percentile(differences, 2.5)
    ci_upper = np.percentile(differences, 97.5)
    
    return ci_lower, ci_upper
```

## Вспомогательные метрики — ARPU, ARPPU, AOV

````python
# Общая подготовка данных (один раз для всех метрик)
orders_with_group = orders_AB.copy()
orders_with_group['revenue'] = (orders_with_group['total_price'] * 0.05).round(2)
orders_with_group = orders_with_group.merge(
    users_with_groups[['user_id', 'group']], 
    on='user_id', 
    how='inner'
)

# Общие агрегаты по группам
revenue_by_group = orders_with_group.groupby('group')['revenue'].sum().reset_index()
revenue_by_group.columns = ['group', 'total_revenue']

group_sizes = users_with_groups.groupby('group').size().reset_index()
group_sizes.columns = ['group', 'total_users']

paying_users_by_group = orders_with_group.groupby('group')['user_id'].nunique().reset_index()
paying_users_by_group.columns = ['group', 'paying_users']

orders_count_by_group = orders_with_group.groupby('group').size().reset_index()
orders_count_by_group.columns = ['group', 'total_orders']

# ARPU
arpu_table = revenue_by_group.merge(group_sizes, on='group')
arpu_table['ARPU'] = (arpu_table['total_revenue'] / arpu_table['total_users']).round(4)

arpu_control = arpu_table[arpu_table['group'] == 'control']['ARPU'].values[0]
arpu_treatment = arpu_table[arpu_table['group'] == 'treatment']['ARPU'].values[0]

arpu_abs_change = arpu_treatment - arpu_control
arpu_rel_change = ((arpu_treatment - arpu_control) / arpu_control * 100) if arpu_control != 0 else 0

# ARPPU
arppu_table = revenue_by_group.merge(paying_users_by_group, on='group')
arppu_table['ARPPU'] = (arppu_table['total_revenue'] / arppu_table['paying_users']).round(4)

arppu_control = arppu_table[arppu_table['group'] == 'control']['ARPPU'].values[0]
arppu_treatment = arppu_table[arppu_table['group'] == 'treatment']['ARPPU'].values[0]

arppu_abs_change = arppu_treatment - arppu_control
arppu_rel_change = ((arppu_treatment - arppu_control) / arppu_control * 100) if arppu_control != 0 else 0

# AOV
aov_table = revenue_by_group.merge(orders_count_by_group, on='group')
aov_table['AOV'] = (aov_table['total_revenue'] / aov_table['total_orders']).round(4)

aov_control = aov_table[aov_table['group'] == 'control']['AOV'].values[0]
aov_treatment = aov_table[aov_table['group'] == 'treatment']['AOV'].values[0]

aov_abs_change = aov_treatment - aov_control
aov_rel_change = ((aov_treatment - aov_control) / aov_control * 100) if aov_control != 0 else 0

# Сводная таблица
rev_control = revenue_by_group[revenue_by_group['group'] == 'control']['total_revenue'].values[0]
rev_treatment = revenue_by_group[revenue_by_group['group'] == 'treatment']['total_revenue'].values[0]

users_control = group_sizes[group_sizes['group'] == 'control']['total_users'].values[0]
users_treatment = group_sizes[group_sizes['group'] == 'treatment']['total_users'].values[0]

paying_control = paying_users_by_group[paying_users_by_group['group'] == 'control']['paying_users'].values[0]
paying_treatment = paying_users_by_group[paying_users_by_group['group'] == 'treatment']['paying_users'].values[0]

orders_control = orders_count_by_group[orders_count_by_group['group'] == 'control']['total_orders'].values[0]
orders_treatment = orders_count_by_group[orders_count_by_group['group'] == 'treatment']['total_orders'].values[0]

summary_data = {
    'Метрика': [
        'Конверсия (CR), %',
        'ARPU, ₽',
        'ARPPU, ₽',
        'AOV, ₽',
        'Общая выручка, ₽',
        'Пользователей всего',
        'Платящих пользователей',
        'Всего заказов'
    ],
    'Control': [
        f"{conv_control:.2f}%",
        f"{arpu_control:.4f}",
        f"{arppu_control:.4f}",
        f"{aov_control:.4f}",
        f"{rev_control:.2f}",
        f"{users_control}",
        f"{paying_control}",
        f"{orders_control}"
    ],
    'Treatment': [
        f"{conv_treatment:.2f}%",
        f"{arpu_treatment:.4f}",
        f"{arppu_treatment:.4f}",
        f"{aov_treatment:.4f}",
        f"{rev_treatment:.2f}",
        f"{users_treatment}",
        f"{paying_treatment}",
        f"{orders_treatment}"
    ],
    'Δ абс.': [
        f"{conv_treatment - conv_control:+.2f} п.п.",
        f"{arpu_abs_change:+.4f}",
        f"{arppu_abs_change:+.4f}",
        f"{aov_abs_change:+.4f}",
        f"{rev_treatment - rev_control:+.2f}",
        f"{users_treatment - users_control:+d}",
        f"{paying_treatment - paying_control:+d}",
        f"{orders_treatment - orders_control:+d}"
    ],
    'Δ отн.': [
        f"{((conv_treatment - conv_control) / conv_control * 100):+.2f}%" if conv_control != 0 else "—",
        f"{arpu_rel_change:+.2f}%" if arpu_control != 0 else "—",
        f"{arppu_rel_change:+.2f}%" if arppu_control != 0 else "—",
        f"{aov_rel_change:+.2f}%" if aov_control != 0 else "—",
        f"{((rev_treatment - rev_control) / rev_control * 100):+.2f}%" if rev_control != 0 else "—",
        f"{((users_treatment - users_control) / users_control * 100):+.2f}%",
        f"{((paying_treatment - paying_control) / paying_control * 100):+.2f}%",
        f"{((orders_treatment - orders_control) / orders_control * 100):+.2f}%"
    ]
}

summary_df = pd.DataFrame(summary_data)

print("ИТОГОВАЯ СВОДКА ВСПОМОГАТЕЛЬНЫХ МЕТРИК A/B ТЕСТА")
display(summary_df.style.set_caption("Сравнение Control vs Treatment"))
````

**Статистическая значимость вспомогательных метрик**

Везде будем использовать тест Mann-Whitney, потому что:

- Распределение дохода не нормальное 
- Маленькие группы платящих (16 в control, 110 в treatment)

````python
 # Статистическая значимость финансовых метрик
from scipy import stats

# Подготовка данных: revenue для каждого пользователя
user_revenue = orders_with_group.groupby('user_id')['revenue'].sum().reset_index()
user_revenue = users_with_groups[['user_id', 'group']].merge(user_revenue, on='user_id', how='left')
user_revenue['revenue'] = user_revenue['revenue'].fillna(0)

# ARPU: все пользователи
arpu_control_values = user_revenue[user_revenue['group'] == 'control']['revenue']
arpu_treatment_values = user_revenue[user_revenue['group'] == 'treatment']['revenue']

stat_arpu, p_arpu = stats.mannwhitneyu(arpu_treatment_values, arpu_control_values, alternative='two-sided')

# ARPPU: только платящие
arppu_control_values = user_revenue[(user_revenue['group'] == 'control') & (user_revenue['revenue'] > 0)]['revenue']
arppu_treatment_values = user_revenue[(user_revenue['group'] == 'treatment') & (user_revenue['revenue'] > 0)]['revenue']

stat_arppu, p_arppu = stats.mannwhitneyu(arppu_treatment_values, arppu_control_values, alternative='two-sided')

# AOV: все заказы (каждый заказ — отдельное наблюдение)
aov_control_values = orders_with_group[orders_with_group['group'] == 'control']['revenue']
aov_treatment_values = orders_with_group[orders_with_group['group'] == 'treatment']['revenue']

stat_aov, p_aov = stats.mannwhitneyu(aov_treatment_values, aov_control_values, alternative='two-sided')

# Вывод результатов
print("Статистическая значимость финансовых метрик (Mann-Whitney U-test)")
print(f"ARPU:  p-value = {p_arpu:.6f}  {'значимо' if p_arpu < 0.05 else 'не значимо'}")
print(f"ARPPU: p-value = {p_arppu:.6f}  {'значимо' if p_arppu < 0.05 else 'не значимо'}")
print(f"AOV:   p-value = {p_aov:.6f}  {'значимо' if p_aov < 0.05 else 'не значимо'}")
````

ARPU — значимо (p < 0.001): бонус увеличил средний доход на пользователя за счёт многократного роста конверсии. 

ARPPU — не значимо (p = 0.61): платящие пользователи в обеих группах приносят одинаково, бонус не повлиял на их ценность. 

AOV — не значимо (p = 0.29): средний чек не изменился, treatment-пользователи не покупают более дешёвые товары. Бонус привлекает больше покупателей, не снижая средний чек и не привлекая «охотников за скидками».
